```python
import nltk

nltk.download('stopwords')
```

```python
stopwords = nltk.corpus.stopwords.words('portuguese')

print(stopwords)
```

```python
from nltk.tokenize import word_tokenize
nltk.download('punkt_tab')
```

```python
frase = 'Eu dirijo devagar porque nós queremos ver os animais'

tokens = word_tokenize(frase)

print(tokens)
```

```python
for t in tokens:
  if t not in stopwords:
    print(t)
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
import pandas as pd

text1 = 'A matématica é muito importante para compreendermos como a natureza funciona'

tf_idf = TfidfVectorizer()
```

```python
vetor = tf_idf.fit_transform([text1])
print(vetor)
```

```python
vetor = vetor.todense()
print(vetor)
```

```python
nomes = tf_idf.get_feature_names_out()

print(nomes)
```

```python
df = pd.DataFrame(vetor, columns=nomes)
print(df)
```



* Corpus: É o conjunto de textos selecionados em um idioma específico para servir
de base de dados. Ele funciona como o material de estudo que o algoritmo usará para analisar, aprender e validar informações.

* Tokenização: É o processo de dividir um texto em partes menores, chamadas de tokens (que geralmente são as palavras e pontuações). É o primeiro passo para o computador conseguir analisar uma frase estruturada.

* Stop Words: São palavras comuns que não trazem significado central ao texto (como "o", "a", "de", "com", "nós"). Elas são removidas logo no início do processamento para tornar o algoritmo mais rápido, sem prejudicar o sentido da frase.

* TF-IDF: É um cálculo estatístico que mede a importância de uma palavra em um
documento. Ele funciona equilibrando duas coisas: dá mais peso para palavras que aparecem várias vezes em um texto (TF), mas reduz o peso se essa palavra for comum demais em todos os outros documentos analisados (IDF). Isso ajuda a destacar as palavras-chave reais de um texto.
